#Estructuras de selección

##if-else
Estructura de seleccion empleada cuando **solo se tienen 2 decisiones posibles**.

Sintaxis
```

if (condicion)
  se ejecuta acciones si y solo si, condicion==v
else:
  se ejecuta si la condicion es falsa

```

````python
x=input("¿Quieres helado? ")
if x.lower()=="sí" or x.lower()=="si":
  h=input("Que sabor quieres?")
  match h:
    case "vainilla": print(f"Aqui tiene su helado de {h}")
    case "Chicle" : print(f"Aqui tiene su helado de {h}")
    case "Ron con pasas" : print(f"Aqui tiene su helado de {h}")
    case _: print("Abrase de aca")
else:
  print("Esta bien :c")
````

##if/else/elif
Estructura de seleccion que se emplea cuando hay mas de 2 opciones disponibles. Se debe colocar tantas condiciones como sea necesario.

```
if (c1): aqui se ejecuta si c1 es verdadera
elif (c2): aqui se ejecuta si c2 es verdadera y c1 es falsa
elif (cn): aqui se ejecuta si cn es verdadera y c1, c2...cn-1 lol, es falsa
else: aqui se ejecuta si todas las condiciones son falsas

##match-case
Estructura de selección, empleada para evaaluar diferentes valores **CON SOLO UNA CONDICIÓN**

Sintaxis
```
match (variable):
  case "valor1": aqui se ejecuta si variable==valor1
  case "valor 2": aqui se ejecuta si variable==valor2
  case "valorn": aqui se ejecuta si variable==valorn
  case _: ninguna de las anteriores
```



##Si ternario

Sintaxis
