#ESTRUCTURA DE REPETICION
Estructura computacional que permite repetir una accion n cantidad de veces o hasta que se logre una condicion.

#Estructura PARA (FOR)
Estructura empleada cuando se sabe cuantas veces se debe repetir una accion

Sintaxis general
```
for (variable) in range (VI, VF, Paso):
Aqui se repite (VF-VI)/Paso veces
```
Donde:
* VI: Valor inicial
* VF: Valor Final
* Paso:

Sintaxis resumida
```
# Cuando no me interesa el valor incial for (variable) in range(VF):
 Se repite VF veces.
```




````python
#Hacer un programa que imprima los numeros pars del 2 al 10
for par in range (2,21,2):
  print(par)
````

````python
for par in range (2,21):
  if par % 2 == 0:
    print(par)
````

````python
print("==============")
print("TABLA DEL 5")
print("==============")
for i in range (1,11):
  multi=5*i
  print(f"5 x {i} = {multi}")
````

````python
print("==============")
print("TABLA DE LOS NUMEROS DEL 1-10")
print("==============")

for n in range (1,11):
  print("-Tabla del", n)

  for i in range (1,11):
    print(f"{n} x {i} = {n*i}")
````

````python
n=int(input("Ingrese su numero: "))
x=0

for i in range (1,n+1):
  if i % 2 != 0:
    x += 1/i
    print(f"+1/{i}")

  else:
    print(f"-1/{i}")
    x -= 1/i
````
