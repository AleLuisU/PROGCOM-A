```python
usuario = input("Ingrese su usuario: ")
password = input("Ingrese su contraseña: ")

especiales = "$@%&#"
tiene_especial = False
tiene_numero = False

for caracter in password:
    if caracter in especiales:
        tiene_especial = True
    if caracter.isdigit():
        tiene_numero = True

if (len(password) >= 10 and
    password[0].isupper() and
    tiene_especial and
    tiene_numero):

    print("Usuario y contraseña creados correctamente")

else:
    print("La contraseña no cumple los requisitos")
```
