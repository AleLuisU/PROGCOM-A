```python
# 3. Hallar el área y perímetro de un círculo.
# IF/elif/else

PI = 3.1416
radio = int(input("Ingrese el radio del circulo: "))
if radio > 0:
  area = PI * radio * radio
  perimetro = 2 * PI * radio
  mensaje = "Calculo valido"
elif radio == 0:
  area = 0
  perimetro = 0
  mensaje = "El radio no puede ser cero"
else:
  area = 0
  perimetro = 0
  mensaje = "El radio no puede ser negativo"
print( "Resultado")
print("Radio:", radio)
print("Area:", area)
print("Perimetro:", perimetro)
print("Mensaje:", mensaje)
# Match case

PI = 3.1416
radio = int(input("Ingrese el radio del circulo: "))
match (radio == 0, radio > 0) :
  case (True, False):
    area = 0
    perimetro = 0
    mensaje = "El radio no puede ser cero"
  case (False, True):
    area = PI * radio * radio
    perimetro = 2 * PI * radio
    mensaje = "Calculo valido"
  case _:
    area = 0
    perimetro = 0
    mensaje = "El radio no puede ser negativo"
print( "Resultado")
print("Radio:", radio)
print("Area:", area)
print("Perimetro:", perimetro)
print("Mensaje:", mensaje)
# Si ternario

PI = 3.1416
radio = int(input("Ingrese el radio del circulo: "))
area , perimetro , mensaje = (PI * radio * radio , 2 * PI * radio , "Calculo valido") if radio > 0 else (0 , 0 , "El radio debe ser mayor a cero")
print( "Resultado")
print("Radio:", radio)
print("Area:", area)
print("Perimetro:", perimetro)
print("Mensaje:", mensaje)
```

```python
# 6.  Hallar el mayor de 3 números.


# 6.if/else/elif
print("Ingresa 2 Números")
v1=float(input("Primer numero: "))
v2=float(input("Segundo numero: "))
v3=float(input("Tercer numero: "))

if v1 > v2:
  if v2 > v3:
    print("El mayor es: {}".format(v1))
  elif v3 > v1:
    print("El mayor es: {}".format(v3))

elif v2 > v1:
  if v2 > v3:
    print("El mayor es: {}".format(v2))
  elif v3 > v2:
    print("El mayor es: {}".format(v3))

# 6.si ternario
print("Ingresa 3 Números")
v1 = float(input("Primer numero: "))
v2 = float(input("Segundo numero: "))
v3 = float(input("Tercer numero: "))

mayor = (v1 if (v1 >= v2 and v1 >= v3) else (v2 if v2 >= v3 else v3))

print(f"El mayor es: {mayor}")
```
